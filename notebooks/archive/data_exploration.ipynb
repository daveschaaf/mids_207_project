```python
# Date range for transactions
print("Date range:")
print(f"Min: {transactions['t_dat'].min()}")
print(f"Max: {transactions['t_dat'].max()}")

# https://www.kaggle.com/c/h-and-m-personalized-fashion-recommendations/discussion/310496
# Prices scaled by factor of 590
jeans = transactions[transactions['article_id'] == 706016001]
print(f"\nJade Skinny Jeans transactions: {len(jeans)}")
print("\nOriginal prices (most common):")
print(jeans['price'].value_counts().head(5))
print("\nPrices × 590:")
print((jeans['price'] * 590).value_counts().head(5))
```
